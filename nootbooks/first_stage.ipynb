```python
import pickle
```

```python
car_dict = {
    "Levy": 1399,
    "Manufacturer": "LEXUS",
    "Model": "RX 450",
    "Prode. year": 2010,
    "Category": "Jeep",
    "Leather interior": "Yes",
    "Fuel type": "Hybrid",
    "Engine volume": 3.5,
    "Mileage": 150000,
    "Cylinders": 6.0,
    "Gear box type": "Automatic",
    "Drive wheels": "4x4",
    "Wheel": "Left wheel",
    "Color": "Silver",
    "Airbags": 12
}
```

# Preprocessing Plan for Model Input

## Input Handling
- Ensure proper handling of input features in
`replace_categorical_by_numerical`:
    - **Engine Volume**: Accept as a numerical input.
    - **Levy**: Input as a number: if not provided, use `0` as the default
    value.
    - **Mileage**: Input as a numerical value.

## Feature Engineering
- Generate a new feature: **`Age`** (e.g., from the car's manufacturing year).

## Feature Removal
- Remove the feature: **`ID`, `Doors`, `Prod. year`** from the dataset.

```python
import pandas as pd
data = pd.DataFrame([car_dict])
data
```

```python
from datetime import datetime

data['Age'] = datetime.now().year - data['Prode. year']

data = data.drop(['Prode. year'], axis=1)
```

```python
data
```

```python
one_hot_columns = ['Leather interior', 'Gear box type', 'Drive wheels', 'Wheel']

with open('../models/one_hot_encoder.pkl', 'rb') as f:
    one_hot_encoder = pickle.load(f)

encoded_data = one_hot_encoder.transform(data[one_hot_columns])
encoded_data_df = pd.DataFrame(
    encoded_data,
    columns=one_hot_encoder.get_feature_names_out(one_hot_columns),
    index=data.index
)

data = pd.concat([data, encoded_data_df], axis=1)
data.drop(columns=one_hot_columns, inplace=True)
```

```python
data
```

```python
label_encode_columns = ['Fuel type', 'Category', 'Model', 'Color', 'Manufacturer']

with open('../models/label_encoders.pkl', 'rb') as f:
    label_encoders = pickle.load(f)
    
for column in label_encode_columns:
    le = label_encoders[column]
    data[column] = le.transform(data[column])
```

```python
data
```

```python
numerical_columns = ['Levy', 'Engine volume', 'Mileage', 'Age']

with open('../models/scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

data[numerical_columns] = scaler.transform(data[numerical_columns])
```

```python
data
```
